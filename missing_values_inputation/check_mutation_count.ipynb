```python
import pandas as pd
```

```python
df = pd.read_csv("../data/METABRIC.csv")
```

```python
mutation_count = df["mutation_count"]
mut_cols = [col for col in df.columns if col[-3:] == "mut"]
mutations = df[mut_cols]

true_mutation_count_list = [(i, 0) for i in range(len(df))]
true_mutation_count = dict(true_mutation_count_list) # {id : true_count}

for i in range(len(mutations)):
    current_patient = mutations.loc[i]
    
    for val in current_patient:
        if val not in [0, "0"]:
            if i in true_mutation_count:
                true_mutation_count[i] += len(str(val).split())
            else:
                true_mutation_count[i] = len(str(val).split())
```

```python
print("TOTAL MISSING VALUES IN MUTATION_COUNT COLUMN :", df["mutation_count"].isnull().sum(), end="\n\n")

nan_count = 0

print("\n  ID" + " "*2 + "MUTATION COUNT" + " "*2 + "TRUE MUTATION COUNT")
for i in range(len(df)):
    if true_mutation_count[i] != df["mutation_count"][i]:
        print(f"{i:>4}" + " "*13 + f"{df["mutation_count"][i]}"  + " "*20 + f"{true_mutation_count[i]}")
        nan_count += 1
        
print("\n\nNUMBER OF VALUES MARKED AS MISSING THAT ARE ACTUALLY EQUAL TO 0 :", nan_count)
```

```python
if nan_count == df["mutation_count"].isnull().sum():
    print("-"*76)
    print("- ALL MISSING VALUES FROM MUTATION COUNT IN DATASET CAN BE REPLACED WITH 0 -")
    print("-"*76)
```

```python
right_count = 0
wrong_count = 0

for i in range(len(df)):
    if df["mutation_count"][i] == true_mutation_count[i]:
        right_count += 1
        continue

    wrong_count += 1
    
if wrong_count == nan_count:
    print("-"*50)
    print("- ALL EXISTING MUTATION COUNT VALUES ARE CORRECT -")
    print("-"*50)
```
