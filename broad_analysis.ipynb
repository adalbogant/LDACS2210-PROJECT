```python
import pandas as pd
```

```python
df = pd.read_csv("data/METABRIC.csv")
print(f"Number of patients in dataset : {len(df)}")
print(f"Number of columns per patient : {len(df.columns)}")
print(f"Total size of dataset : {df.size}")
```

```python
clinical_attributes = df.columns[:31]
mutations = [col for col in df.columns if col[-3:] == "mut"]
z_scores = df.columns[31:len(df.columns)-len(mutations)]

print(f" {len(clinical_attributes)} CLINICAL ATTRIBUTES ".center(50, "-"))
for i, col in enumerate(clinical_attributes):
    print(f"{i:>3} : {col}")

print("\n" + f" {len(z_scores)} MRNA Z-SCORE VALUES ".center(50, "-"))
for i, col in enumerate(z_scores):
    print(f"{(i+len(clinical_attributes)):>3} : {col}")
    
print("\n" + f" {len(mutations)} GENE MUTATIONS ".center(50, "-"))
for i, col in enumerate(mutations):
    print(f"{(i+len(clinical_attributes)+len(z_scores)):>3} : {col}")
```

```python
missing = (df.isnull().sum()).tolist()
missing_non_zero = [df.columns[i] for i in range(len(missing)) if missing[i] != 0]

for i, col in enumerate(missing):
    if col != 0:
        print(f"{df.columns[i]:<30} : {col:>3} values = {(col / len(df) * 100):>2.0f} %")
```

```python
from sklearn.manifold import TSNE, MDS
from sklearn.preprocessing import StandardScaler
import plotly.express as px

# Appliquer reduction dimension que à genes expression ; colorer selon ...
mRNA_zscores = df.iloc[:, range(31, 520)]

# Standardisation does not have a great impact, as values are already more or less standard (by nature)
# TODO : check if no problem of scaling by line and not columns
scaler = StandardScaler()
standard_mRNA_zscores = scaler.fit_transform(mRNA_zscores)
print("Means : ", scaler.mean_)
print("Variances : ", scaler.var_)


tsne_test = TSNE()
tsne_standard = TSNE()
reduced_mRNA_zscores = tsne_test.fit_transform(mRNA_zscores)
tsne_standard_zscores = tsne_standard.fit_transform(standard_mRNA_zscores)
```

```python
fig = px.scatter(x=reduced_mRNA_zscores[:, 0], y=reduced_mRNA_zscores[:, 1], color=df["nottingham_prognostic_index"])
fig_standard = px.scatter(x=tsne_standard_zscores[:, 0], y=tsne_standard_zscores[:, 1], color=df["neoplasm_histologic_grade"])
fig.show()
fig_standard.show()
```

```python
from sklearn.decomposition import PCA

pca_test = PCA(n_components=2)
pca_standard = PCA(n_components=2)
# Appliquer reduction dimension que à genes expression ; colorer selon ...
pca_mRNA_zscores = pca_test.fit_transform(mRNA_zscores)
pca_standard_mRNA_zscores = pca_standard.fit_transform(standard_mRNA_zscores)
print(pca_test.explained_variance_ratio_)
```

```python
fig = px.scatter(x=pca_mRNA_zscores[:, 0], y=pca_mRNA_zscores[:, 1], color=df["nottingham_prognostic_index"])
fig_standard_pca = px.scatter(x=pca_standard_mRNA_zscores[:, 0], y=pca_standard_mRNA_zscores[:, 1], color=df["cancer_type_detailed"])
fig.show()
fig_standard_pca.show()
```
